```python
import numpy as np
import matplotlib.pyplot as plt
import ipywidgets as widget
from ipywidgets import interact

def puente_monofasico(Freq=50,Amplitud=1,Ciclos=3):
    if Freq>=10:
        fs=Freq*1000
    else:
        fs=10000
    T=1/Freq    #periodo
    t=np.arange(0,T,1/fs) #vector de tiempo de un periodo
    v_in=Amplitud*np.sin(2*np.pi*Freq*t)
    v_out=np.where(v_in>=0,v_in,0)
    
    #repeticion de multiples ciclos para la señal
    v_in_total=np.tile(v_in,Ciclos)
    v_out_total=np.tile(v_out,Ciclos)
    t_total=np.linspace(0,Ciclos*T,len(v_in_total))

    #graficacion de resultados
    plt.figure(figsize=(10,6))
    plt.plot(t_total,v_out_total,'b',label='Salida')
    plt.plot(t_total,v_in_total,'r--',label='Entrada')
    plt.legend()
    plt.grid()

interact(puente_monofasico,
         Freq=widget.IntSlider(value=5,min=5,max=1000,step=10),
         Amplitud=widget.IntSlider(value=1,min=0.5,max=5,step=0.5),
         Ciclos=widget.IntSlider(value=3,min=1,max=10,step=1))

```
